```python
# Q1. Find the Frequency of Each Word in a Sentence
def frequency(text):
    words=text.split()
    result={}
    for i in words:
        if i in result:
            result[i]+=1
        else:
            result[i]=1
    return result
print(frequency("python is easy and python is powerful"))
```

```python
# Q2.Find the Frequency of Each Character in a Word Write a Python program to count how many times each character appears in a given word and store the result in a dictionary.
def frequent(text):
    freq={}
    for i in text:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    return freq
print(frequent("banana"))
```

```python
#Q3. Find the First Non-Repeating Character in a Word
def least(text):
    freq={}
    for i in text:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    return min(freq,key=freq.get)
print(least("swiss"))
```

```python
#Q4. Check Whether Two Strings Are Anagrams
def anagrams(s1,s2):
    return sorted(s1)==sorted(s2)
print(anagrams("listen","silent"))
```

```python
#Q5. Find the First Repeating Character in a String
def frequent(text):
    letter=set()
    for i in text:
        if i in letter:
            return i
        letter.add(i)
    return None
print(frequent("programming"))
```

```python
#Q6. Move All Zeros to the End of a List
def move_zero(li,target):
    non_zero=[]
    zeros=[]
    for i in li:
        if i==target:
            zeros.append(i)
        else:
            non_zero.append(i)
    return non_zero+zeros
print(move_zero([0, 3, 0, 1, 2, 0, 4],0))
```

```python
#Q7. Find the Second Highest Distinct Number in a List
def second_highest(li):
    li=sorted(set(li))
    return li[-2]
print(second_highest([10, 20, 40, 30, 40, 20]))
```

```python
#Q8. Remove Duplicates Without Changing the Original Order

def duplicates(li):
    result=[]
    for i in li:
        if i not in result:
            result.append(i)
    return result
print(duplicates( [4, 2, 4, 5, 2, 7, 5]))
```

```python
#Q9. Find the Missing Number from 1 to N

def missing(li,N):
    for i in range(1,N):
        if i not in li:
            return i
    return None
print(missing([1, 2, 4, 5, 6],6))
```

```python
#Q10. Find Two Numbers Whose Sum Equals a Target

def sum_two(nums,target):
    result={}
    for i in nums:
        complement = target - i
        if complement in result:
            return [complement, i]
        result[i] = True
    return[]
print(sum_two([2, 7, 11, 15], 9))
```

```python
#Q11. Find the Number Closest to Zero

def close_zero(nums):
    close=nums[0]
    for i in nums:
        if abs(i)<abs(close):
            close=i
        elif abs(i)==abs(close) and abs(i)>abs(close):
            close=i
    return close
print(close_zero([-8, -3, 2, 5, -2]))
```

```python
#Q12. Find Common Elements Between Two Lists

def common(l1,l2):
    result=[]
    for i in l1:
        if i in l2 and i not in result:
            result.append(i)
    return result
print(common([1, 2, 3, 4, 5],[3, 4, 5, 6, 7]))
```

```python
#Q13. Rotate a List to the Right by K Positions
def rotate_right(numbers, k):
    if not numbers:
        return []
    k %= len(numbers)
    return numbers[-k:] + numbers[:-k] if k else numbers[:]
print(rotate_right( [1, 2, 3, 4, 5], 2))
```

```python
#Q14. Find the Maximum Sum of Two Consecutive Elements
def max_sum_of_consecutive(numbers):
    if len(numbers) < 2:
        return None
    max_sum = numbers[0] + numbers[1]
    for i in range(1, len(numbers) - 1):
        current_sum = numbers[i] + numbers[i + 1]
        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum
print(max_sum_of_consecutive( [4, 7, 2, 9, 5]))
```

```python
#Q15. Find All Pairs of Numbers Whose Sum Equals a Target
def all_pairs_sum(numbers, target):
    pairs = []
    seen = set()
    for current in numbers:
        needed = target - current
        if needed in seen:
            pairs.append((needed, current))
        seen.add(current)
    return pairs
print(all_pairs_sum([2, 4, 3, 5, 7, 8, 1], 9))
```
