Q1. Find the Shortest Word in a Sentence

Write a Python program to find the shortest word in a given sentence.

Example input:
Python makes coding easy

Expected output:
easy

```python
text=input()
def shortest(text):
    words=text.split()
    small=words[0]
    for word in words:
        if len(word)<len(small):
            small=word
    return small
print(shortest(text))
```

Q6. Count Vowels in a String

Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.

Example input:
Education

Expected output:
5

```python
word=input()
def vowel(word):
    count=0
    for i in word:
        if i in 'aeiouAEIOU':
            count+=1
    return count
print(vowel(word))
```

Q7. Count Consonants in a String

Example input:

Q10. Find All Characters Appearing More Than Once

Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.

Example input:
programming

Expected output:
['r', 'g', 'm']

```python
text=input()
def frequent(text):
    freq={}
    for i in text:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    result=[]
    for i in text:
        if i!=' ' and freq[i]>1 and i not in result:
            result.append(i)
    return result
print(frequent(text))
```

Q15. Create a Dictionary of Numbers and Their Even/Odd Status

Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.

Example input:
[1, 2, 3, 4, 5]

Expected output:
{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}


```python
num=eval(input())
def evod(num):
    nums={}
    for i in num:
        if i%2==0:
            nums[i]='even'
        else:
            nums[i]='odd'
    return nums
print(evod(num))
```

Q16. Find the Most Frequent Word in a Sentence

Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.

Example input:
Python is easy and Python is powerful

Expected output:
python

```python
text=input()
def frequent(text):
    words=text.lower().split()
    freq={}
    for i in words:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    most_freq=max(freq,key=freq.get)
    return most_freq
print(frequent(text))
```

Q18. Find the First Word That Appears Only Once

Write a Python program to find the first word that occurs exactly once in a sentence. Treat uppercase and lowercase versions as the same word.

apple banana apple mango banana orange

Expected output:

```python
text=input()
def first(text):
    words=text.split()
    freq={}
    for i in words:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    for i in words:
        if freq[i]==1:
            return i
print(first(text))
```
